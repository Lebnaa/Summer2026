```python
# step 1
import numpy as np

w0, w1, w2 = -1.5, 0.5, 1.0

def sigmoid(s):
    return 1 / (1 + np.exp(-s))

x1_s1, x2_s1 = 0.6, 2.1
s1 = w0*1 + w1*x1_s1 + w2*x2_s1
y_hat1 = sigmoid(s1)

x1_s2, x2_s2 = 2.0, 1.0
s2 = w0*1 + w1*x1_s2 + w2*x2_s2
y_hat2 = sigmoid(s2)

print('QUESTION 1: Outputs')
print(f's(1) = {s1}')
print(f'y_hat(1) = {y_hat1:.6f}')
print(f's(2) = {s2}')
print(f'y_hat(2) = {y_hat2:.6f}')
```

```python
# Step 2
y1_true, y2_true = 2.0, -1.0

mse = (1/2) * ((y_hat1 - y1_true)**2 + (y_hat2 - y2_true)**2)

print('QUESTION 2: MSE Loss')
print(f'MSE = {mse:.6f}')
```

```python
# Step 3
def sigmoid_deriv(y_hat):
    return y_hat * (1 - y_hat)

delta1 = (y_hat1 - y1_true) * sigmoid_deriv(y_hat1)
delta2 = (y_hat2 - y2_true) * sigmoid_deriv(y_hat2)

dJ_dw0 = (1/2) * (delta1*1     + delta2*1)
dJ_dw1 = (1/2) * (delta1*x1_s1 + delta2*x1_s2)
dJ_dw2 = (1/2) * (delta1*x2_s1 + delta2*x2_s2)

print('QUESTION 3: Gradient dJ/dw')
print(f'dJ/dw0 = {dJ_dw0:.6f}')
print(f'dJ/dw1 = {dJ_dw1:.6f}')
print(f'dJ/dw2 = {dJ_dw2:.6f}')
```
